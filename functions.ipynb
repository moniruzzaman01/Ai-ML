```python
def pattern(num):
    for n in range(num):
        for i in range(n):
            print("*",end="")
        print()
pattern(10)
```

# Unknown numbers of argument accept

```python
def square_sum(*arg):
    sum =0
    for a in arg:
        sum+=(a*a)
    return sum
square_sum(2,2,2,2,2)
```

# Key value pair type argument send and accept

```python
def key_arg(**arg):
    print(arg)

key_arg(name='shakib',age='26')
```

```python
def dict_print(dct):
    print(dct)

dict_print({'name':'shakib','age':26})
```

# Lamda Function - in js we called anonymous function

```python
func = lambda a,b:a+b
print(func(3,4))
```

# map

```python
lst = [1,2,3,4,5]

sqr = map(lambda x: x*x if x%2==0 else x*x*x,lst)
print(list(sqr))
```

# filter

```python
lst = [x for x in range(1,101)]
filtered_lst = filter(lambda x: x%2==0, lst )
print(list(filtered_lst))
```
