```python
import numpy as np
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/128.0.0.0 Safari/537.36"
}
res = requests.get('https://en.wikipedia.org/wiki/List_of_companies_of_Bangladesh',headers=headers).text
```

```python
soup = BeautifulSoup(res)
table = soup.find('table')
trs = table.find_all('tr')
```

```python
company = []
industry = []
sector = []
headquarter = []
founded = []
for tr in trs[2:]:
    tds = tr.find_all('td')
    company.append(tds[0].text.strip())
    industry.append(tds[1].text.strip())
    sector.append(tds[2].text.strip())
    headquarter.append(tds[3].text.strip())
    founded.append(tds[4].text.strip())

dct = dict({'company':company, 'industry':industry,'sector':sector,'headquarter':headquarter,'founded':founded})

df = pd.DataFrame(dct)
df.head()
```

```python
# df.loc[df['sector']=='-', 'sector']=np.nan
df['sector']=df['sector'].replace('-',np.nan)
display(df.head(50))
display(df.info())
```

```python
newdf = df.dropna()
display(newdf.info())
display(newdf.head(50))
```
