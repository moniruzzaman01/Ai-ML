```python
def safe_divide(a,b):
    try:
        return a/b
    except:
        return "Division by zero"

print(safe_divide(20,5))
print(safe_divide(20,0))
```

```python
class Shape:
    def area(self):
        return 0
class Rectangle(Shape):
    def __init__(self,w,h):
        self.w=w
        self.h=h 
    def area(self):
        print(f"Area: {(self.w*self.h):.2f}")
class Circle(Shape):
    def __init__(self,r):
        self.r=r
    def area(self):
        print(f"Area: {(3.1416*self.r*self.r):.2f}")
        

input_list = input().split()
name=input_list[0]

if name=='Rectangle':
    w=float(input_list[1])
    h=float(input_list[2])
    rect = Rectangle(w,h)
    rect.area()
elif name=='Circle':
    r= float(input_list[1])
    cir = Circle(r)
    cir.area()
        
```

```python
def find_factorial(n):
    if n==0 or n==1:
        return 1
    return n * find_factorial(n-1)

number_str = input()
number = int(number_str)

total=0
while number/10:
    total+=find_factorial(number%10)
    number=int(number/10)

print("Strong Number" if total==int(number_str) else "Not Strong Number")
```
