```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
import plotly.express as px
```

```python
students1 = pd.read_csv('students1.csv')
sns.scatterplot(data=students1, x='study_hours', y='test_score', hue='gender')
```

```python

sns.scatterplot(data=students1, x='study_hours', y='test_score', hue='gender', style='class_level')
```

```python

sns.scatterplot(data=students1, x='study_hours', y='test_score', hue='gender', style='subject', size='Tshirt_size')
```

```python

sns.relplot(kind='scatter', data=students1, x='study_hours', y='test_score', hue='gender', style='subject', size='Tshirt_size')
```

```python
students = pd.read_csv('students.csv')
students.head()
```
