```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
import plotly.express as px
```

```python
students = pd.read_csv('students.csv')
students_mark = students[['Marks','study_hours','attendance_rate','gender']]
students_mark.head()
```

```python
fig = px.histogram(students_mark, x='attendance_rate')
fig.show()
```

```python
fig = px.histogram(students_mark, x='attendance_rate',color='gender')
fig.show()
```
