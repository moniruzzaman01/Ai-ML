```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
import plotly.express as px
```

```python
students = pd.read_csv('students.csv')
students.head()
```

```python
sns.relplot(kind='scatter', data=students, x='study_hours', y='test_score', col='gender')
```

```python
sns.relplot(kind='scatter', data=students, x='study_hours', y='test_score', col='gender', row='hostel')
```

```python
sns.relplot(kind='scatter', data=students, x='study_hours', y='test_score', col='week')
```

```python
sns.relplot(kind='scatter', data=students, x='study_hours', y='test_score', col='week', col_wrap=4)
```

### Histogram using Seaborn

#### Axes level

```python
sns.histplot(data=students, x='attendance_rate', hue='gender', bins=10)
```

#### Figure Level

```python
sns.displot(data=students, x='attendance_rate', col='gender')
```

```python
sns.displot(data=students, x='study_hours', col='gender')
```
