# Line Plot
#### Axes Level

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
import plotly.express as px
```

```python
students = pd.read_csv('students1.csv')
sns.lineplot(data=students, x='week', y='attendance_rate', errorbar=None, color='orange')
```

```python
sns.set_style("whitegrid") # or "darkgrid"
sns.lineplot(data=students, x='week', y='attendance_rate', errorbar=None, color='orange',hue='gender')
```

```python
plot = sns.lineplot(data=students, x='week', y='attendance_rate', errorbar=None, color='orange',hue='gender')
plot.grid(True, color='gray', linestyle='-.', linewidth=0.5)
```

#### Figure Level

```python
plot = sns.relplot(kind='line', data=students, x='week', y='attendance_rate', errorbar=None, color='orange',hue='class_level', )
```
