Inheritence Attribute

```python
class Phone:
    category = "Electronics"
    
    #constructor
    def __init__(self,brand,price):
            self.brand = brand
            self.price = price

apple = Phone('apple',200000)
print(apple.brand)
```

Methods

```python
class Phone:
    category = "Electronics"
    
    #constructor
    def __init__(self,brand,price):
            self.brand = brand
            self.price = price
    #Methods
    def printBrand(self):
        print(f"Brand is: {self.brand}");

apple = Phone('apple',200000)
apple.printBrand()
samsung = Phone('samsung',200000)
samsung.printBrand()
```

Inheritence - Single

```python
# Base class / Parent class
class PersonalInfo:
    def __init__(self,name,age,gender):
            self.name=name
            self.age=age
            self.gender=gender
            
    #Methods 
    def printName(self):
        print(f"Name from parent: {self.name}")
        
# Inherited class / Child class
class Student(PersonalInfo):
    def __init__ (self,name,age,gender):
        super().__init__(name,age,gender)
    
    def printName(self):
        super().printName()
        print(f"Name from child: {self.name}")


class Student1(PersonalInfo):
    def __init__ (self,name,age,gender):
        PersonalInfo.__init__(self,name,age,gender)
    
    def printName(self):
        PersonalInfo.printName(self)
        print(f"Name from child1: {self.name}")



shakib = Student1('shakib',26,'male')
print(shakib.printName())
```

Inheritence - Multiple

```python
# Base class / Parent class
class PersonalInfo:
    def __init__(self,name,age,gender):
            self.name=name
            self.age=age
            self.gender=gender
            
    #Methods 
    def printName(self):
        print(f"Name from parent: {self.name}")
        
# Base class / Parent class
class Address:
    def __init__(self,builidng_no,road_no,area):
            self.builidng_no=builidng_no
            self.road_no=road_no
            self.area=area
    #Methods
    def printAddress(self):
        print(f"building no: {self.builidng_no}, road no: {self.road_no}, area: {self.area}")
            
        
# Inherited class / Child class
class Student(PersonalInfo,Address):
    def __init__(self,name,age,gender,building_no,road_no,area):
        PersonalInfo.__init__(self,name,age,gender)
        Address.__init__(self,building_no,road_no, area)
    
    #Methods 
    def printName(self):
        PersonalInfo.printName(self)
        print("Hey I am name from children")
        
    def printAddress(self):
        Address.printAddress(self)
        print("Hey I am address from children")
        

shakib = Student('shakib',26,'male',10,54,'Gazipur')
print(shakib.printAddress())
```

```python
class Student:
    def __init__(self,stdId,name):
        self.__stdId=stdId
        self.__name=name
        
    def getId(self):
        print(self.__stdId)
    def getName(self):
        return self.__name
    def setName(self,name):
        self.__name=name

shakib = Student('001','shakib')
print(shakib.getName())
shakib.setName('shakib al hasan')
print(shakib.getName())
```
