```python
import pandas as pd
import matplotlib.pyplot as plt

users = pd.read_json('users_info.json')
grouped_users = users.groupby('gender').size()
plt.xlabel("Gender")
plt.ylabel("Frequency")
plt.bar(grouped_users.index,grouped_users.values)
plt.show()
```

```python
students = pd.read_csv('student_data.csv')
grouped_students = students.groupby('CompletionStatus').size()
plt.xlabel("Completion Status")
plt.ylabel("Frequency")
plt.bar(grouped_students.index,grouped_students.values, color=['green','orange','red'], edgecolor='black')
plt.show()
```
