```python
import pandas as pd
df = pd.read_json('datas/students_data.json')
df
```

```python
def generate_marks(marks):
    if(pd.isna(marks)):
        return 0
    result = 0
    a = int(marks%10)
    if(a>5):
        x = int(marks/10)
        result=(x*10)+10
    else:
        y = int(marks/10)
        result = y*10
    return result
    
df['New DS Marks']=df['Data Structure Marks'].apply(generate_marks)
df
```
