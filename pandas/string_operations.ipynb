```python
import pandas as pd
data = {
  "name": ["Alice", "Bob", "Charlie", "Diana", "Ethan", "Fiona", "George", "Hannah",],
  "city": ["New York", "Los Angeles", "Chicago", "Houston", "Miami", "Seattle", "Boston New", "Denver"],
  "department": ["HR", "IT", "Finance", "Marketing", "IT", "HR", "Operations", "Finance"],
  "salary": [5500, 7200, 6800, 6100, 7500, 5300, 7000, 6600]
}
df = pd.DataFrame(data)
df
```

```python
# Letter A exists or not
df.loc[df['city'].str.contains('New')]
```

```python
# starts with New word
df.loc[df['city'].str.contains('^New')]
```

```python
# Ends with New word
df.loc[df['city'].str.contains('New$')]
```

```python
# stop case sensitivity
df.loc[df['city'].str.contains('e',case=False)]
```
