# Nd Array From Scratch

```python
# zeros - zeros_like

import numpy as np
arr = np.zeros(5)
# print(arr)
arr1 = np.zeros((5))
# print(arr1)
arr2 = np.zeros((5,2))
# print(arr2)
arr3 = np.zeros((5,3,2))
# print(arr3)


old_array = [
    [
        [1,2,3],
        [1,2,3],
        [1,2,3],
    ]
]
arr4 = np.zeros_like(old_array)
print(arr4)
```

```python
# ones - ones_like

import numpy as np
arr = np.ones(5)
# print(arr)
arr1 = np.ones((5))
# print(arr1)
arr2 = np.ones((5,2))
# print(arr2)
arr3 = np.ones((5,3,2))
print(arr3)


old_array = [
    [
        [1,2,3],
        [1,2,3],
        [1,2,3],
    ]
]
arr4 = np.ones_like(old_array)
# print(arr4)
```

```python
# empty - empty_like

import numpy as np
arr = np.empty(5)
print(arr)
arr1 = np.empty((5))
# print(arr1)
arr2 = np.empty((5,2))
# print(arr2)
arr3 = np.empty((5,3,2))
# print(arr3)


old_array = [
    [
        [1,2,3],
        [1,2,3],
        [1,2,3],
    ]
]
arr4 = np.empty_like(old_array)
# print(arr4)
```

```python
# full - full_like

import numpy as np
arr = np.full(5,1)
# print(arr)
arr1 = np.full((5,1),5)
# print(arr1)
arr2 = np.full((5,2),4)
# print(arr2)
arr3 = np.full((5,3,2),9)
# print(arr3)


old_array = [
    [
        [1,2,3],
        [1,2,3],
        [1,2,3],
    ]
]
arr4 = np.full_like(old_array,78)
print(arr4)
```

# Matrix Slicing

```python
import numpy as np
arr = np.array([
    [5,6,7,8],
    [1,2,3,4],
])
lst = []
for i in np.nditer(arr):
    lst.append(i)
# print(np.array(lst))

lst1=[]
for i in arr:
    for j in i:
        lst1.append(j)
# print(np.array(lst))

arr1 = np.arange(10,100,5).reshape(3,6)
print(arr1)
print('------------------')
reshaped_arr1 = arr1[:2]
print(reshaped_arr1)
print('------------------')
reshaped_arr2 = arr1[:,1:4]
print(reshaped_arr2)
print('------------------')
# reshaped_arr3 = arr1[row start:row end,col start:col end]
reshaped_arr3 = arr1[1:2,1:4]
print(reshaped_arr3)
```
