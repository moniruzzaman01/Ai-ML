```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame({
    'h':[150,160,170,180,190],
    'w':[50,60,70,80,90]
})
display(df)
```

```python
mn = df['h'].mean()
std = df['h'].std()
print(mn,std)
df['standardization'] = (df['h']-mn)/std
display(df)
```

```python
df['standardization'].plot(kind='bar')
plt.title('Z Scaling Features')
plt.xlabel('Index') 
plt.ylabel('Z Values')
plt.show()
```

```python
df.drop(columns='standardization', inplace=True)
mn = df.mean()
sd = df.std()
z = (df-mn)/sd
display(z)
```

```python
z.plot(kind='bar')
plt.title('Z Scaling Features')
plt.xlabel('Index') 
plt.ylabel('Z Values')
plt.show()
```

```python
mn = df.min()
mx = df.max()
x = (df-mn)/(mx-mn)
display(x)
```

```python
df = pd.DataFrame({
    'name':['alex','hales','john','steve','maria','joye','helena'],
    'color':['red','blue','green','yellow','green','red','red'],
    'size':['small','medium','large','small','large','large','medium'],
    'price':[20,50,30,10,50,80,90],
})
display(df)
```

```python
d_color = pd.get_dummies(df['color'], prefix='C', dtype=int)
display(d_color)
```

```python
newdf = pd.concat([df,d_color], axis=1)
newdf
```

```python
newdf = newdf.drop(columns='color', axis=1)
newdf
```

```python
order = {'small':1, 'medium':2, 'large':3}
newdf['encoded_size'] = newdf['size'].map(order).astype(int)
newdf
```
