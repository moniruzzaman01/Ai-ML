```python
from functools import reduce
with open('datas/sample.txt','r') as file:
    content = file.readlines()
    lines = len(content) # line count
    words = reduce(lambda x,y:x+y,list(map(lambda x: len(x.split()),content))) # word count
    # find characters
    content_without_enter = list(map(lambda x:x.strip(),content))
    content_without_space = list(map(lambda x:x.replace(" ",""),content_without_enter))
    character_count_list = list(map(lambda x:len(x),content_without_space))
    character_count = reduce(lambda x,y:x+y,character_count_list)
    print(character_count)
    
with open('datas/output.txt','w') as file:
    file.write(f"count- {lines}\n")
    file.write(f"words- {words}\n")
    file.write(f"characters- {character_count}\n")
```
